```python
import os
import torch
import torchvision
import torchvision.transforms as transforms
from torch.utils.data import DataLoader, random_split
```

```python
data_dir = r"C:\Users\User\OneDrive\Documents\GitHub\AI-ML-assessment-cifar10-project\data"
transform = transforms.Compose([
    transforms.Resize((32, 32)),
    transforms.ToTensor(),
    transforms.Normalize((0.5, 0.5, 0.5), (0.5, 0.5, 0.5))
])
```

```python
train_val_dataset = torchvision.datasets.CIFAR10(root='./data', train=True, download=True, transform=transform)
test_dataset = torchvision.datasets.CIFAR10(root='./data', train=False, download=True, transform=transform)

train_size = 45000
val_size = 5000
train_dataset, val_dataset = random_split(train_val_dataset, [train_size, val_size])
```

```python
train_loader = DataLoader(train_dataset, batch_size=64, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=64, shuffle=False)
test_loader = DataLoader(test_dataset, batch_size=64, shuffle=False)

print(f"Classes: {train_val_dataset.classes}")
print(f"Training samples: {len(train_dataset)}")
print(f"Validation samples: {len(val_dataset)}")
print(f"Test samples: {len(test_dataset)}")
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
import matplotlib.pyplot as plt
```

```python
class SimpleMLP(nn.Module):
    def __init__(self, input_dim=32*32*3, hidden_dim=512, output_dim=10):
        super(SimpleMLP, self).__init__()
        self.flatten = nn.Flatten()
        self.fc1 = nn.Linear(input_dim, hidden_dim)
        self.relu = nn.ReLU()
        self.dropout = nn.Dropout(0.2)
        self.fc2 = nn.Linear(hidden_dim, 256)
        self.fc3 = nn.Linear(256, output_dim)

    def forward(self, x):
        x = self.flatten(x)
        x = self.relu(self.fc1(x))
        x = self.dropout(x)
        x = self.relu(self.fc2(x))
        x = self.fc3(x)
        return x

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = SimpleMLP().to(device)
```

```python
total_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f"Total Trainable Parameters: {total_params:,}")

criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)
```

```python
num_epochs = 10
history = {'train_loss': [], 'train_acc': [], 'val_loss': [], 'val_acc': []}
```

```python
for epoch in range(num_epochs):
    model.train()
    running_loss = 0.0
    correct = 0
    total = 0

    for inputs, labels in train_loader:
        inputs, labels = inputs.to(device), labels.to(device)

        optimizer.zero_grad()
        outputs = model(inputs)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

        running_loss += loss.item() * inputs.size(0)
        _, predicted = torch.max(outputs.data, 1)
        total += labels.size(0)
        correct += (predicted == labels).sum().item()

    epoch_loss = running_loss / len(train_loader.dataset)
    epoch_acc = correct / total
    history['train_loss'].append(epoch_loss)
    history['train_acc'].append(epoch_acc)

    model.eval()
    val_loss = 0.0
    val_correct = 0
    val_total = 0

    with torch.no_grad():
        for inputs, labels in val_loader:
            inputs, labels = inputs.to(device), labels.to(device)
            outputs = model(inputs)
            loss = criterion(outputs, labels)

            val_loss += loss.item() * inputs.size(0)
            _, predicted = torch.max(outputs.data, 1)
            val_total += labels.size(0)
            val_correct += (predicted == labels).sum().item()

    val_epoch_loss = val_loss / len(val_loader.dataset)
    val_epoch_acc = val_correct / val_total
    history['val_loss'].append(val_epoch_loss)
    history['val_acc'].append(val_epoch_acc)

    print(f"Epoch [{epoch+1}/{num_epochs}], "
          f"Train Loss: {epoch_loss:.4f}, Train Acc: {epoch_acc:.4f}, "
          f"Val Loss: {val_epoch_loss:.4f}, Val Acc: {val_epoch_acc:.4f}")
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(history['train_loss'], label='Train Loss')
plt.plot(history['val_loss'], label='Val Loss')
plt.title('Loss across Epochs')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(history['train_acc'], label='Train Accuracy')
plt.plot(history['val_acc'], label='Val Accuracy')
plt.title('Accuracy across Epochs')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()

plt.show()
```

## 1. Introduction and dataset preparation

CIFAR-10 contains 32?32 RGB images from ten classes. The 50,000 official training images were divided into 45,000 training examples and 5,000 validation examples. Images were converted to tensors and each channel was normalized with mean 0.5 and standard deviation 0.5.

## 2. Model architecture and training

`SimpleMLP` flattens each image into 3,072 input features, then applies fully connected layers of 512, 256, and 10 units. ReLU activations follow the first two layers, with dropout (`p=0.2`) after the 512-unit layer. The model has 1,707,274 trainable parameters and was trained for 10 epochs using Adam with learning rate 0.001 and cross-entropy loss.

## 3. Results and evaluation

Training accuracy increased from 40.58% in epoch 1 to 63.19% in epoch 10. Validation accuracy reached its highest recorded value, 53.08%, in epoch 8, then ended at 52.56%. Training loss continued to fall, from 1.6806 to 1.0464, while validation loss reached 1.3733 in epoch 8 and rose to 1.4327 by epoch 10. The later divergence between training and validation trends is consistent with overfitting.

The classification report gives 53% accuracy on the 10,000-image official test set (rounded in the displayed report). Class recall was lowest for dog (36%), deer (38%), and cat (44%); it was highest for ship (67%), automobile (65%), and frog (64%). The report also shows lower precision for cat (34%) and bird (41%). These results indicate that the MLP had difficulty separating several visually varied CIFAR-10 classes.

## 4. Conclusion and limitations

The MLP learned useful class patterns, but its final validation accuracy was 52.56% and its official test accuracy was reported as 53%. Flattening removes the images' spatial layout before classification, so the network does not directly exploit nearby pixels the way a CNN does. This is a plausible architectural limitation, though the results alone do not establish the cause of each error.

The training/validation split was created with `random_split` without a fixed generator. Therefore, the saved metrics describe this run's split and are not guaranteed to reproduce on another run. The test report is the final evaluation for this run; it should not be used to tune the model.
